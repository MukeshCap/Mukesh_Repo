```python
%sql

CREATE TABLE IF NOT EXISTS dev.configs.ingestion_config (
  config_id         BIGINT GENERATED ALWAYS AS IDENTITY (START WITH 1 INCREMENT BY 1),
  source_type       STRING  NOT NULL COMMENT 'Where the source lives, e.g. file, adls, s3, database',
  source_name       STRING  NOT NULL COMMENT 'Logical name of the source dataset',
  source_folder     STRING  NOT NULL COMMENT 'Landing path / volume path of the source files',
  ingestion_method  STRING  NOT NULL COMMENT 'autoloader | lakeflow_pipeline | declarative_pipeline',
  source_format     STRING  NOT NULL COMMENT 'csv | parquet | json',
  is_active         BOOLEAN NOT NULL COMMENT 'TRUE = ingestion for this table is enabled',
  target_schema     STRING  NOT NULL COMMENT 'Target schema for the ingested table',
  target_table      STRING  NOT NULL COMMENT 'Target table name',
  load_type         STRING  NOT NULL COMMENT 'incremental | overwrite',
  target_catalog STRING NOT NULL ,
  checkpoint_base STRING
  
  
)
;

INSERT INTO dev.configs.ingestion_config
  (source_type, source_name, source_folder, ingestion_method, source_format, is_active, target_schema, target_table, load_type,target_catalog,checkpoint_base)
VALUES
  ('file', 'adls_files',    'abfss://indiaprojectraw@indiaprojectstorage.dfs.core.windows.net/customer/',    'autoloader',           'json',  TRUE,  '1_bronze', 'customer',    'incremental','dev','/Volumes/dev/configs/checkpoints'),
  ('file', 'adls_files',    'abfss://indiaprojectraw@indiaprojectstorage.dfs.core.windows.net/customeraddress/',    'autoloader',           'csv',  TRUE,  '1_bronze', 'customeraddress',    'incremental','dev','/Volumes/dev/configs/checkpoints');

-- 5. Check the rows
SELECT * FROM dev.configs.ingestion_config ORDER BY config_id;
```
