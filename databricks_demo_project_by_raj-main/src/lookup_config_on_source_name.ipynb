```python
import json

source_name = dbutils.widgets.get("source_name")

df = spark.table("dev.configs.ingestion_config")

filtered_df = df.filter(df["source_name"] == source_name).filter(df["is_active"] == True)

config_records = [row.asDict() for row in filtered_df.collect()]

print(f"Found {len(config_records)} record(s) for source_name='{source_name}'")
print(config_records)
```

```python
dbutils.jobs.taskValues.set(key="config_records", value=config_records)

```
