```python
from pyspark.sql.functions import *

# 10 million rows
df = spark.range(10000000) \
          .withColumn("country", expr("""
              case
                  when id % 3 = 0 then 'India'
                  when id % 3 = 1 then 'USA'
                  else 'UK'
              end
          """))

df.count()
```

```python
# Cache the dataframe in memory
df.cache()

# Materialize cache
df.count()
```

```python
df.groupBy("country").count().collect()
```
