REPARTITION

```python
spark.conf.set("spark.databricks.io.cache.enabled", "false")
```

```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("PartitionDemo").getOrCreate()

# Create 1000 records
data = [(i, f"Employee_{i}", i % 10) for i in range(1, 10000000)]

df = spark.createDataFrame(data, ["emp_id", "emp_name", "dept_id"])

df.show(5)
print("Total Records:", df.count())
```

```python
print(df.rdd.getNumPartitions())
```

```python
df.spark_partition_id().show()
```

```python
from pyspark.sql.functions import spark_partition_id

df.groupBy(spark_partition_id().alias("partition_id")) \
  .count() \
  .orderBy("partition_id") \
  .show()
```

```python
partition_sizes = df.rdd.mapPartitions(lambda x: [sum(1 for _ in x)]).collect()

print(partition_sizes)
```

```python
df = df.repartition(20)
```

```python
partition_sizes = df.rdd.mapPartitions(lambda x: [sum(1 for _ in x)]).collect()

print(partition_sizes)
```

COLASCE

```python
spark.conf.set("spark.databricks.io.cache.enabled", "false")
```

```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("PartitionDemo").getOrCreate()

# Create 1000 records
data = [(i, f"Employee_{i}", i % 10) for i in range(1, 10000000)]

df = spark.createDataFrame(data, ["emp_id", "emp_name", "dept_id"])

df.show(5)
print("Total Records:", df.count())
```

```python
print(df.rdd.getNumPartitions())
```

```python
partition_sizes = df.rdd.mapPartitions(lambda x: [sum(1 for _ in x)]).collect()

print(partition_sizes)
```

```python
df = df.coalesce(2)
```

```python
partition_sizes = df.rdd.mapPartitions(lambda x: [sum(1 for _ in x)]).collect()

print(partition_sizes)
```
