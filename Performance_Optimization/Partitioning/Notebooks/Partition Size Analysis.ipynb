```python
spark.conf.set("spark.databricks.io.cache.enabled", "false")
```

```python
df_repartition = spark.read.format('json')\
    .load('/Volumes/dbacademy_retail/v02/business_daily_events')
```

```python
print(df_repartition.rdd.getNumPartitions())
```

```python
from pyspark.sql.functions import spark_partition_id, count

partition_sizes = (df_repartition
    .groupBy(spark_partition_id().alias("partition_id"))
    .agg(count("*").alias("size"))
    .orderBy("partition_id")
    .collect())

print([row["size"] for row in partition_sizes])
```

```python
df_repartition = df_repartition.repartition(2)
```

```python
from pyspark.sql.functions import spark_partition_id, count

partition_sizes = (df_repartition
    .groupBy(spark_partition_id().alias("partition_id"))
    .agg(count("*").alias("size"))
    .orderBy("partition_id")
    .collect())

print([row["size"] for row in partition_sizes])
```
