```python
import requests
import json

# Groww public API endpoints for stock/mutual fund data
BASE_URL = "https://groww.in/v1/api"

def get_stock_data(search_query):
 """Search for stocks on Groww and return basic details."""
 url = f"{BASE_URL}/search/v1/entity"
 params = {"app": "false", "entity_type": "stock", "page": 0, "q": search_query, "size": 10}
 headers = {"User-Agent": "Mozilla/5.0"}
 response = requests.get(url, params=params, headers=headers)
 response.raise_for_status()
 return response.json()

def get_top_gainers():
 """Fetch top gaining stocks."""
 url = f"{BASE_URL}/stocks_data/v1/tr_liv_prices/exchange/NSE/top_gainers"
 headers = {"User-Agent": "Mozilla/5.0"}
 response = requests.get(url, headers=headers)
 response.raise_for_status()
 return response.json()

def get_top_losers():
 """Fetch top losing stocks."""
 url = f"{BASE_URL}/stocks_data/v1/tr_liv_prices/exchange/NSE/top_losers"
 headers = {"User-Agent": "Mozilla/5.0"}
 response = requests.get(url, headers=headers)
 response.raise_for_status()
 return response.json()

def get_mutual_fund_data(search_query):
 """Search for mutual funds on Groww."""
 url = f"{BASE_URL}/search/v1/entity"
 params = {"app": "false", "entity_type": "mutual_fund", "page": 0, "q": search_query, "size": 10}
 headers = {"User-Agent": "Mozilla/5.0"}
 response = requests.get(url, params=params, headers=headers)
 response.raise_for_status()
 return response.json()

def get_stock_details(groww_contract_id):
 """Get detailed stock info by Groww contract ID (e.g., 'RELIANCE')."""
 url = f"{BASE_URL}/stocks_data/v1/accord_points/exchange/NSE/segment/CASH/{groww_contract_id}/latest"
 headers = {"User-Agent": "Mozilla/5.0"}
 response = requests.get(url, headers=headers)
 response.raise_for_status()
 return response.json()

# --- Example Usage ---
print("=== Searching for 'Reliance' stocks ===")
try:
 results = get_stock_data("Reliance")
 print(json.dumps(results, indent=2))
except Exception as e:
 print(f"Error fetching stock search results: {e}")

print("\n=== Top Gainers (NSE) ===")
try:
 gainers = get_top_gainers()
 for stock in gainers.get("topGainersDataDtoList", [])[:5]:
 print(f"  {stock.get('companyName', 'N/A')} | LTP: {stock.get('ltp', 'N/A')} | Change: {stock.get('dayChangePerc', 'N/A')}%")
except Exception as e:
 print(f"Error fetching top gainers: {e}")

print("\n=== Top Losers (NSE) ===")
try:
 losers = get_top_losers()
 for stock in losers.get("topLosersDataDtoList", [])[:5]:
 print(f"  {stock.get('companyName', 'N/A')} | LTP: {stock.get('ltp', 'N/A')} | Change: {stock.get('dayChangePerc', 'N/A')}%")
except Exception as e:
 print(f"Error fetching top losers: {e}")

print("\n=== Searching for 'SBI' mutual funds ===")
try:
 mf_results = get_mutual_fund_data("SBI")
 print(json.dumps(mf_results, indent=2))
except Exception as e:
 print(f"Error fetching mutual fund results: {e}")
```
