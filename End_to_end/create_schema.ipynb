```python
data = [
    (1, "John Smith", "North"),
    (2, "Mike Johnson", "South"),
    (3, "David Brown", "East")
]

df = spark.createDataFrame(
    data,
    ["agent_id", "agent_name", "region"]
)

# Create a single csv file
df.coalesce(1) \
  .write \
  .mode("overwrite") \
  .option("header", "true") \
  .csv("/Volumes/uat/mukesh/agent/temp")

files = dbutils.fs.ls("/Volumes/uat/mukesh/agent/temp")

csv_file = [f.path for f in files if f.name.startswith("part-")][0]

dbutils.fs.cp(
    csv_file,
    "/Volumes/uat/mukesh/agent/agent.csv"
)

dbutils.fs.rm("/Volumes/uat/mukesh/agent/temp", recurse=True)
```
