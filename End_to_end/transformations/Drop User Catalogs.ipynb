```python
# Fetch all catalogs in the metastore
catalogs = spark.sql("SHOW CATALOGS").collect()

for row in catalogs:
    catalog_name = row['catalog']
    
    # Skip system/default catalogs that cannot or should not be dropped
    # 'samples' is a Databricks-owned catalog (owner: System user) and cannot be dropped
    if catalog_name not in ['hive_metastore', 'system', 'main', 'samples']:
        print(f"Dropping catalog: {catalog_name}")
        # CASCADE ensures all schemas and tables inside the catalog are also dropped
        spark.sql(f"DROP CATALOG IF EXISTS {catalog_name} CASCADE")
```
